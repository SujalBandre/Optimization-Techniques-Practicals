```python
import sympy
from sympy import symbols, diff, Matrix, solve

print("--- Finding and Classifying Stationary Points of a Multivariable Function ---")
print("Using Partial Derivatives and Hessian Matrix")

# 1. Define the multivariable function
# Let's use an example function f(x, y) = x^3 + y^3 - 3xy
# You can change this to your desired function.

x, y = symbols('x y')
f = x**2 + y**2 - 4*x - 6*y + 13

print(f"\nOriginal function f(x, y): {f}")

# 2. Calculate First-Order Partial Derivatives
# These are needed to find the critical points.

df_dx = diff(f, x)
df_dy = diff(f, y)

print(f"\nFirst partial derivative w.r.t x (df/dx): {df_dx}")
print(f"First partial derivative w.r.t y (df/dy): {df_dy}")

# 3. Find Critical Points
# Set the first partial derivatives to zero and solve the system of equations.
# solve() can return different formats depending on the solution type.

print("\nSolving df/dx = 0 and df/dy = 0 to find critical points...")
critical_points_raw = solve([df_dx, df_dy], (x, y))

# Ensure critical_points is always a list of dictionaries for consistent iteration
if isinstance(critical_points_raw, dict):
    critical_points = [critical_points_raw]
else:
    critical_points = critical_points_raw

print(f"Critical points: {critical_points}")

# 4. Calculate Second-Order Partial Derivatives
# These are used to form the Hessian matrix.

d2f_dx2 = diff(df_dx, x) # f_xx
d2f_dy2 = diff(df_dy, y) # f_yy
d2f_dxdy = diff(df_dx, y) # f_xy
d2f_dydx = diff(df_dy, x) # f_yx (should be equal to f_xy for most functions)

print(f"\nSecond partial derivative f_xx: {d2f_dx2}")
print(f"Second partial derivative f_yy: {d2f_dy2}")
print(f"Second partial derivative f_xy: {d2f_dxdy}")
print(f"Second partial derivative f_yx: {d2f_dydx}")

# 5. Form the Hessian Matrix
# H = [[f_xx, f_xy],
#      [f_yx, f_yy]]
H = Matrix([[d2f_dx2, d2f_dxdy],
            [d2f_dydx, d2f_dy2]])

print(f"\nHessian Matrix H:\n{H}")

# 6. Classify Critical Points using the Hessian Determinant (Second Derivative Test)
print("\n--- Classifying Critical Points ---")

for point_solution_dict in critical_points:
    x_val = point_solution_dict[x]
    y_val = point_solution_dict[y]

    # Evaluate the Hessian matrix at each critical point
    hessian_at_point = H.subs({x: x_val, y: y_val})

    # Calculate the determinant of the Hessian (D)
    D = hessian_at_point.det()

    # Evaluate f_xx at the critical point
    f_xx_at_point = d2f_dx2.subs({x: x_val, y: y_val})

    print(f"\nFor critical point (x={x_val}, y={y_val}):")
    print(f"  Hessian Matrix at point:\n{hessian_at_point}")
    print(f"  Determinant of Hessian (D): {D}")
    print(f"  f_xx at point: {f_xx_at_point}")

    if D > 0:
        if f_xx_at_point > 0:
            print("  -> Local Minimum")
        elif f_xx_at_point < 0:
            print("  -> Local Maximum")
        else:
            print("  -> Test is inconclusive (f_xx = 0, D > 0)")
    elif D < 0:
        print("  -> Saddle Point")
    elif D == 0:
        print("  -> Test is inconclusive (D = 0), further analysis is required.")
    else:
        print("  -> Could not classify (D is not a number)")
```

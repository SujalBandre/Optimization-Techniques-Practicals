```python
import sympy as sp

# 1. Define the variable and the function
x = sp.symbols('x')
f = 2*x**3 - 9*x**2 + 12*x

print(f"Original Function f(x) = {f}\n")

# 2. First Derivative Test: Find critical points
f_prime = sp.diff(f, x)
critical_points = sp.solve(f_prime, x)

print(f"First Derivative f'(x) = {f_prime}")
print(f"Critical Points: {critical_points}\n")

# 3. Second Derivative Test: Classify critical points
f_double_prime = sp.diff(f_prime, x)
print(f"Second Derivative f''(x) = {f_double_prime}\n")

print("--- Classification Results ---")
for point in critical_points:
    # Evaluate the second derivative at the critical point
    test_value = f_double_prime.subs(x, point)

    # Get the actual y-value of the function at this point
    y_value = f.subs(x, point)

    if test_value < 0:
        print(f"At x = {point}: f''(x) = {test_value} (< 0) -> Local MAXIMUM at ({point}, {y_value})")
    elif test_value > 0:
        print(f"At x = {point}: f''(x) = {test_value} (> 0) -> Local MINIMUM at ({point}, {y_value})")
    else:
        print(f"At x = {point}: f''(x) = 0 -> Test Inconclusive (Check first derivative sign changes)")
```

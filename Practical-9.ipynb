```python
import numpy as np
from scipy.optimize import milp, Bounds, LinearConstraint

# Objective: Maximize 5x + 4y  ==>  Minimize -5x - 4y
c = np.array([-5.0, -4.0])

# Variable types: both x and y are Integers (1 indicates integer)
integrality = np.array([1, 1])

# Bounds: x >= 0, y >= 0
bounds = Bounds(lb=[0.0, 0.0], ub=[np.inf, np.inf])

# Constraints:
# 6x + 4y <= 24
# x + 2y <= 6
A = np.array([
    [6.0, 4.0],
    [1.0, 2.0]
])

# Constraints lower bounds (None represented as -inf) and upper bounds
constraints = LinearConstraint(A, lb=[-np.inf, -np.inf], ub=[24.0, 6.0])

# Solve using scipy's mixed-integer linear programming solver
res = milp(c=c, integrality=integrality, bounds=bounds, constraints=constraints)

if res.success:
    print("Status: Optimal")
    print(f"Optimal Value of x = {int(round(res.x[0]))}")
    print(f"Optimal Value of y = {int(round(res.x[1]))}")
    print(f"Maximum Profit = {int(round(-res.fun))}")
else:
    print("Optimization failed:", res.message)
```

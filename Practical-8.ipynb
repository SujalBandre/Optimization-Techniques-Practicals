```python
INF = float('inf')
graph = [
 [0, 3, INF, 7],
 [8, 0, 2, INF],
 [5, INF, 0, 1],
 [2, INF, INF, 0]
]
n = len(graph)
dist = [row[:] for row in graph]
for k in range(n):
    for i in range(n):
        for j in range(n):
            dist[i][j] = min(
                dist[i][j],
                dist[i][k] + dist[k][j]
            )
print("Shortest Distance Matrix:")
for row in dist:
    print(row)
```
